```python
#learned how to do this from google
import random
import urllib.request
import json

# LEADERBOARD

leaderboard = []


# FUNCTIONS


def get_word_from_api():
    """Gets a random word from an online API."""
#If no internet
    backup_words = [
        "computer",
        "python",
        "school",
        "cheese",
        "science",
        "soccer",
        "pizza",
        "programming"
    ]

    try:
        # Get a random word from the API
        url = "https://random-word-api.herokuapp.com/word?number=1"

        response = urllib.request.urlopen(url, timeout=5)

        data = json.loads(response.read().decode())

        word = data[0].lower()

        # Make sure the word is usable
        if word.isalpha() and len(word) >= 3:
            return word

        return random.choice(backup_words)

    except Exception:
        print("Chatbot: I couldn't connect to the word API.")
        print("Chatbot: I'll use a backup word instead.")

        return random.choice(backup_words)


def choose_difficulty():
    """Choose difficulty and number of guesses."""

    print()
    print("Chatbot: Choose a difficulty:")
    print("1 = Easy - 5 guesses")
    print("2 = Medium - 7 guesses")
    print("3 = Hard - 9 guesses")

    while True:
        try:
            difficulty = input("You: Enter 1, 2, or 3: ")

            if difficulty not in ("1", "2", "3"):
                raise ValueError

            break

        except ValueError:
            print("Chatbot: Please enter 1, 2, or 3.")

    if difficulty == "1":
        guesses = 5

    elif difficulty == "2":
        guesses = 7

    else:
        guesses = 9

    return guesses


def give_hint(word, guesses):
    """Give the first two letters using string slicing."""

    first_two = word[0:2]

    print("Chatbot: Hint! The word starts with:", first_two)

    guesses -= 2

    if guesses < 0:
        guesses = 0

    print("Chatbot: The hint cost you 2 guesses.")

    return guesses


def display_word(word, guessed_letters):
    """Display guessed letters and blanks."""

    display = ""

    for letter in word:

        if letter in guessed_letters:
            display += letter + " "

        else:
            display += "- "

    return display


def word_is_complete(word, guessed_letters):
    """Check if every letter has been guessed."""

    for letter in word:

        if letter not in guessed_letters:
            return False

    return True


def calculate_points(guesses, used_hint):
    """Calculate the player's score."""

    points = guesses * 10

    # Bonus for having guesses left
    if guesses >= 5:
        points += 25

    # Hint penalty
    if used_hint:
        points -= 20

    if points < 0:
        points = 0

    return points


def show_leaderboard(scores):
    """Display the leaderboard from highest to lowest."""


    print("          LEADERBOARD")


    scores.sort(key=lambda x: x[1], reverse=True)

    for position, player in enumerate(scores, 1):
        print(
            str(position) + ".",
            player[0],
            "-",
            player[1],
            "points"
        )




# WELCOME


print("Chatbot: Welcome to the Word Guessing Game!")


# DIFFICULTY

guesses = choose_difficulty()


# GET WORD FROM API

print()
print("Chatbot: Getting a random word...")
#also learned this from google searching
word = get_word_from_api()

print("Chatbot: I have chosen a word!")
print("Chatbot: You have", guesses, "guesses.")


# GAME VARIABLES
#

guessed_letters = []

used_hint = False

game_won = False

game_lost = False


# MAIN GAME LOOP

while guesses > 0:

    print()

    display = display_word(word, guessed_letters)

    print("Chatbot:", display)

    print("Chatbot: Guesses remaining:", guesses)

    # Check if the word is complete
    if word_is_complete(word, guessed_letters):

        game_won = True
        break

    # Get player's guess
    guess = input(
        "You: Enter a letter, 'hint', or 'solve': "
    ).lower().strip()


    # WHOLE WORD SOLVE

    if guess == "solve":

        full_guess = input(
            "You: Enter your whole-word guess: "
        ).lower().strip()

        if full_guess == word:

            print()
            print("Chatbot: Congratulations!")
            print("Chatbot: You guessed the whole word!")
            print("Chatbot: The word was:", word)

            game_won = True

        else:

            print()
            print("Chatbot: Wrong!")
            print("Chatbot: Your whole-word guess was incorrect.")
            print("Chatbot: The word was:", word)

            # Wrong solve immediately ends the game
            game_lost = True

        break


    # HINT

    if guess == "hint":

        if guesses >= 2:

            guesses = give_hint(word, guesses)

            used_hint = True

        else:

            print(
                "Chatbot: You need at least 2 guesses "
                "to use a hint."
            )

        continue


    # VALIDATE LETTER

    if len(guess) != 1 or not guess.isalpha():

        print("Chatbot: Please enter exactly one letter.")

        continue


    # REPEATED LETTER

    if guess in guessed_letters:

        print("Chatbot: You already guessed that letter.")

        # Repeated guesses still cost 1 guess
        guesses -= 1

        print(
            "Chatbot: Repeated guesses cost 1 guess."
        )

        continue


    # NEW LETTER

    guessed_letters.append(guess)

    # Every letter guess costs 1 guess
    guesses -= 1


    # CHECK LETTER

    if guess in word:

        print(
            "Chatbot: Yes! That letter is in the word."
        )

    else:

        print(
            "Chatbot: Sorry, that letter is not in the word."
        )


    # CHECK WIN

    if word_is_complete(word, guessed_letters):

        game_won = True
        break


# GAME END


if game_won:

    print("Chatbot: Congratulations!")
    print("Chatbot: You guessed the word:", word)

    # Calculate score
    points = calculate_points(
        guesses,
        used_hint
    )

    # Bonus for winning
    points += 25

    print("Chatbot: Your score is", points, "points!")


elif game_lost:

    print("Chatbot: Game over!")
    print("Chatbot: The word was:", word)

    points = 0


else:

    print("Chatbot: Game over!")
    print("Chatbot: You ran out of guesses.")
    print("Chatbot: The word was:", word)

    points = 0


# LEADERBOARD

print()

player_name = input(
    "Chatbot: Enter your name for the leaderboard: "
)

# Add player's score
leaderboard.append(
    (player_name, points)
)

# Show leaderboard
show_leaderboard(leaderboard)



print("Chatbot: Thanks for playing!")
```
